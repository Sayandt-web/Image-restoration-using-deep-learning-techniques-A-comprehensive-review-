```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

processed_images = []
image_names = []

for image_path in uploaded.keys():

    # Read image
    img = cv2.imread(image_path)

    # Convert BGR → RGB
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Resize
    img = cv2.resize(img, (256, 256))

    # Normalize
    img = img.astype(np.float32) / 255.0

    processed_images.append(img)
    image_names.append(image_path)

print("Total Images Loaded:", len(processed_images))
print("Shape of First Image:", processed_images[0].shape)
```

```python
import math

n = len(processed_images)

cols = 3
rows = math.ceil(n / cols)

plt.figure(figsize=(15, 10))

for i, img in enumerate(processed_images):

    plt.subplot(rows, cols, i+1)
    plt.imshow(img)
    plt.title(image_names[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import torch

image_tensors = []

for img in processed_images:

    tensor = torch.tensor(
        img,
        dtype=torch.float32
    ).permute(2,0,1).unsqueeze(0)

    image_tensors.append(tensor)

print("Number of Tensors:", len(image_tensors))
print("Tensor Shape:", image_tensors[0].shape)
```

```python
import torch.nn as nn

class RDN(nn.Module):

    def __init__(self):
        super(RDN, self).__init__()

        self.network = nn.Sequential(

            nn.Conv2d(3,64,3,padding=1),
            nn.ReLU(),

            nn.Conv2d(64,64,3,padding=1),
            nn.ReLU(),

            nn.Conv2d(64,64,3,padding=1),
            nn.ReLU(),

            nn.Conv2d(64,3,3,padding=1)

        )

    def forward(self,x):
        return self.network(x)

rdn = RDN()

rdn.eval()

print("RDN Loaded Successfully")
```

```python
denoised_images = []

with torch.no_grad():

    for tensor in image_tensors:

        output = rdn(tensor)

        output = output.squeeze().permute(1,2,0).numpy()

        output = np.clip(output,0,1)

        denoised_images.append(output)

print("Denoised Images:", len(denoised_images))
```

```python
import torch
import torch.nn as nn

class ChannelAttention(nn.Module):

    def __init__(self, channels):
        super().__init__()

        self.avg_pool = nn.AdaptiveAvgPool2d(1)

        self.fc = nn.Sequential(
            nn.Conv2d(channels, channels // 16, 1),
            nn.ReLU(),
            nn.Conv2d(channels // 16, channels, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        weight = self.fc(self.avg_pool(x))
        return x * weight


class RCAN(nn.Module):

    def __init__(self):
        super().__init__()

        self.conv1 = nn.Conv2d(3, 64, 3, padding=1)

        self.ca = ChannelAttention(64)

        self.conv2 = nn.Conv2d(64, 3, 3, padding=1)

    def forward(self, x):

        x = self.conv1(x)
        x = self.ca(x)
        x = self.conv2(x)

        return x

rcan = RCAN()
rcan.eval()

print("RCAN Loaded Successfully")
```

```python
enhance_tensors = []

for img in denoised_images:

    tensor = torch.tensor(
        img,
        dtype=torch.float32
    ).permute(2,0,1).unsqueeze(0)

    enhance_tensors.append(tensor)

print("Total Tensors:", len(enhance_tensors))
```

```python
import numpy as np

enhanced_images = []

with torch.no_grad():

    for tensor in enhance_tensors:

        output = rcan(tensor)

        output = output.squeeze().permute(1,2,0).numpy()

        output = np.clip(output,0,1)

        # normalize for visualization
        output = (output - output.min()) / (
            output.max() - output.min() + 1e-8
        )

        enhanced_images.append(output)

print("Enhanced Images:", len(enhanced_images))
```

```python
import cv2
import numpy as np

enhanced_images = []

for img in denoised_images:

    img_uint8 = (img * 255).astype(np.uint8)

    # White balance
    lab = cv2.cvtColor(img_uint8, cv2.COLOR_RGB2LAB)

    l, a, b = cv2.split(lab)

    clahe = cv2.createCLAHE(
        clipLimit=3.0,
        tileGridSize=(8,8)
    )

    l = clahe.apply(l)

    enhanced_lab = cv2.merge((l, a, b))

    enhanced = cv2.cvtColor(
        enhanced_lab,
        cv2.COLOR_LAB2RGB
    )

    enhanced = enhanced.astype(np.float32) / 255.0

    enhanced_images.append(enhanced)

print("Enhancement Completed")
```

```python
import cv2
import numpy as np

denoised_images = []

for img in processed_images:

    img_uint8 = (img * 255).astype(np.uint8)

    denoised = cv2.fastNlMeansDenoisingColored(
        img_uint8,
        None,
        10,  # h
        10,  # hColor
        7,
        21
    )

    denoised = denoised.astype(np.float32) / 255.0

    denoised_images.append(denoised)

print("Denoising Complete")
```

```python
from skimage.metrics import peak_signal_noise_ratio
from skimage.metrics import structural_similarity
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
import numpy as np

# Lists to store results
psnr_scores = []
ssim_scores = []
mse_scores = []
mae_scores = []

for original, enhanced in zip(processed_images, enhanced_images):

    # Convert to uint8
    original_uint8 = (original * 255).astype(np.uint8)
    enhanced_uint8 = (enhanced * 255).astype(np.uint8)

    # PSNR
    psnr = peak_signal_noise_ratio(
        original_uint8,
        enhanced_uint8,
        data_range=255
    )

    # SSIM
    ssim = structural_similarity(
        original_uint8,
        enhanced_uint8,
        channel_axis=2,
        data_range=255
    )

    # MSE
    mse = mean_squared_error(
        original_uint8.flatten(),
        enhanced_uint8.flatten()
    )

    # MAE
    mae = mean_absolute_error(
        original_uint8.flatten(),
        enhanced_uint8.flatten()
    )

    psnr_scores.append(psnr)
    ssim_scores.append(ssim)
    mse_scores.append(mse)
    mae_scores.append(mae)

# Average Results
print("Average PSNR :", np.mean(psnr_scores))
print("Average SSIM :", np.mean(ssim_scores))
print("Average MSE  :", np.mean(mse_scores))
print("Average MAE  :", np.mean(mae_scores))
```

```python
import pandas as pd

results = pd.DataFrame({
    "Image": [f"Image_{i+1}" for i in range(len(psnr_scores))],
    "PSNR": psnr_scores,
    "SSIM": ssim_scores,
    "MSE": mse_scores,
    "MAE": mae_scores
})

print(results)
```
